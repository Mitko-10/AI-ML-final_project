Advanced Analysis of Dijkstra's Shortest Path Algorithm1. Introduction and Historical ContextDijkstra's algorithm was designed by computer scientist Edsger W. Dijkstra in 1956 and published in 1959. It serves as a solution to the Single-Source Shortest Path (SSSP) problem for a weighted graph. Beyond its practical applications in GPS and network routing, it is a masterpiece of greedy algorithmic design.2. Formal Mathematical FrameworkWe represent the problem environment as a weighted directed graph $G$. Formally, the system is defined by the following 4-tuple:$$G = \langle V, E, w, s \rangle$$Where each component is defined as:$V$: A finite, non-empty set of vertices (nodes), $V = \{v_1, v_2, \dots, v_n\}$.$E$: A set of edges, where $E \subseteq V \times V$. An edge $e = (u, v)$ represents a directed transition from node $u$ to node $v$.$w$: A weight function $w: E \to \mathbb{R}_{\geq 0}$. For every edge $(u, v) \in E$, $w(u, v)$ is a non-negative real number representing the cost of transition.$s$: The source vertex, $s \in V$, where the search originates.3. Path Theory and Optimal SubstructureTo analyze the algorithm's correctness, we define a path $p$ as a sequence of vertices $\langle v_0, v_1, \dots, v_k \rangle$. The weight of the path $p$ is given by:$$w(p) = \sum_{i=1}^{k} w(v_{i-1}, v_i)$$The shortest-path distance from $s$ to $v$ is then defined as:$$\delta(s, v) = \begin{cases} \min \{w(p) : s \xrightarrow{p} v\} & \text{if a path exists} \\ \infty & \text{otherwise} \end{cases}$$Optimal Substructure Lemma:Dijkstra’s algorithm relies on the principle that if $p = \langle v_1, \dots, v_i, \dots, v_j, \dots, v_k \rangle$ is a shortest path from $v_1$ to $v_k$, then the subpath $p_{ij}$ is also the shortest path between $v_i$ and $v_j$. This allows for the recursive-like greedy build-up of the solution.4. The Shortest Path Tree (SPT)During execution, the algorithm constructs a predecessor subgraph $G_{\pi} = (V, E_{\pi})$. Upon termination, $G_{\pi}$ is a Shortest Path Tree, which satisfies:$G_{\pi}$ is a rooted tree with root $s$.For every $v \in V$, the unique path from $s$ to $v$ in $G_{\pi}$ is a shortest path in $G$.The edges $E_{\pi}$ are defined by the predecessors $prev[v]$ updated during relaxation.5. Algorithmic Logic and InvariantsThe algorithm maintains a set $S$ of vertices whose final shortest-path weights $\delta(s, v)$ have already been determined.The Relaxation Step:For each edge $(u, v)$, the algorithm maintains an upper bound $dist[v]$ on the weight of the shortest path. The Relaxation Lemma states that for an edge $(u, v)$:$$\text{RELAX}(u, v, w):$$$$\text{if } dist[v] > dist[u] + w(u, v):$$$$dist[v] \leftarrow dist[u] + w(u, v)$$$$prev[v] \leftarrow u$$Triangle Inequality Invariant:For all edges $(u, v) \in E$, the final distances must satisfy:$$\delta(s, v) \leq \delta(s, u) + w(u, v)$$6.
